```python
from dataclasses import dataclass, make_dataclass, asdict
```

```python
@dataclass
class C:
    a: int       # 'a' has no default value
    b: int = 0   # assign a default value for 'b'
```

```python
def make_paths(out_names, out_paths) -> dataclass:
    return make_dataclass(cls_name="StorageCatalog", fields=[(o, str) for o in out_names ])(*out_paths)
```

```python
class Nb():
    def _paths(self, outs: list[str]):
        root_path = "s3://backet/path/"
        return [root_path + out_name for out_name in outs] 

    def __init__(self, outs: list[str]):
        self.outs = make_paths(outs, self._paths(outs=outs))
```

```python
nb1 = Nb(["a", "b", "c"])
nb2 = Nb(["a", "d"])
```

```python
print(nb1.outs.a)
print(nb2.outs.d)
```
